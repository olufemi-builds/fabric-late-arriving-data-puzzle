```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window
from pyspark.sql.types import (
    StructType, StructField, StringType, TimestampType,
    DecimalType, IntegerType, BooleanType
)
from datetime import datetime

print("PySpark imports completed successfully.")
```

```python
# 1. Convert customer modified_at to timestamp
wave1_customers = (
    wave1_customers_raw
    .withColumn(
        "modified_at",
        F.to_timestamp("modified_at", "yyyy-MM-dd HH:mm:ss")
    )
)

# 2. Convert order_date, amount and modified_at
wave1_orders = (
    wave1_orders_raw
    .withColumn(
        "order_date",
        F.to_date("order_date", "yyyy-MM-dd")
    )
    .withColumn(
        "amount",
        F.col("amount").cast(DecimalType(18, 2))
    )
    .withColumn(
        "modified_at",
        F.to_timestamp("modified_at", "yyyy-MM-dd HH:mm:ss")
    )
)

# 3. Verify the data types
print("CUSTOMER SCHEMA")
wave1_customers.printSchema()

print("ORDER SCHEMA")
wave1_orders.printSchema()
```

```python
from pyspark.sql import functions as F

# Check missing values in customer data
print("WAVE 1 CUSTOMERS — MISSING VALUES")

customer_nulls = wave1_customers.select([
    F.count(
        F.when(
            F.col(c).isNull() | (F.trim(F.col(c)) == ""),
            c
        )
    ).alias(c)
    for c in wave1_customers.columns
])

display(customer_nulls)


# Check missing values in order data
print("WAVE 1 ORDERS — MISSING VALUES")

order_nulls = wave1_orders.select([
    F.count(
        F.when(
            F.col(c).isNull() | (F.trim(F.col(c)) == ""),
            c
        )
    ).alias(c)
    for c in wave1_orders.columns
])

display(order_nulls)
```

```python
# 1. Apply customer attribute defaults
wave1_customers_clean = (
    wave1_customers
    .withColumn(
        "city",
        F.when(
            F.col("city").isNull() | (F.trim(F.col("city")) == ""),
            "Unknown"
        ).otherwise(F.trim(F.col("city")))
    )
    .withColumn(
        "province",
        F.when(
            F.col("province").isNull() | (F.trim(F.col("province")) == ""),
            "Unknown"
        ).otherwise(F.trim(F.col("province")))
    )
    .withColumn(
        "loyalty_tier",
        F.when(
            F.col("loyalty_tier").isNull() | (F.trim(F.col("loyalty_tier")) == ""),
            "Bronze"
        ).otherwise(F.trim(F.col("loyalty_tier")))
    )
)

# 2. Identify orders with blank or missing amounts
rejected_orders_w1 = wave1_orders.filter(
    F.col("amount").isNull()
)

# 3. Keep orders with valid amounts
valid_orders_w1 = wave1_orders.filter(
    F.col("amount").isNotNull()
)

# 4. Map blank customer IDs to the unknown member
valid_orders_w1 = valid_orders_w1.withColumn(
    "customer_id",
    F.when(
        F.col("customer_id").isNull() |
        (F.trim(F.col("customer_id")) == ""),
        "-1"
    ).otherwise(F.trim(F.col("customer_id")))
)

# 5. Check the results
print("CUSTOMERS AFTER CLEANING:", wave1_customers_clean.count())
print("VALID ORDERS:", valid_orders_w1.count())
print("REJECTED ORDERS:", rejected_orders_w1.count())

print("CUSTOMER DEFAULTS")
display(
    wave1_customers_clean
    .filter(
        (F.col("city") == "Unknown") |
        (F.col("province") == "Unknown") |
        (F.col("loyalty_tier") == "Bronze")
    )
    .limit(10)
)

print("REJECTED ORDERS")
display(rejected_orders_w1.limit(10))
```

```python
# 1. Count exact duplicate customer records
customer_exact_duplicates = (
    wave1_customers_clean.count()
    - wave1_customers_clean.dropDuplicates().count()
)

# 2. Count exact duplicate order records
order_exact_duplicates = (
    valid_orders_w1.count()
    - valid_orders_w1.dropDuplicates().count()
)

# 3. Count repeated customer IDs
repeated_customer_ids = (
    wave1_customers_clean
    .groupBy("customer_id")
    .count()
    .filter(F.col("count") > 1)
)

# 4. Count repeated order IDs
repeated_order_ids = (
    valid_orders_w1
    .groupBy("order_id")
    .count()
    .filter(F.col("count") > 1)
)

print("Exact duplicate customer records:", customer_exact_duplicates)
print("Exact duplicate order records:", order_exact_duplicates)

print("Customer IDs appearing more than once:", repeated_customer_ids.count())
print("Order IDs appearing more than once:", repeated_order_ids.count())

print("Repeated customer IDs:")
display(repeated_customer_ids)

print("Repeated order IDs:")
display(repeated_order_ids)
```

```python
# Remove identical customer records
wave1_customers_dedup = (
    wave1_customers_clean
    .dropDuplicates()
)

# Remove identical valid order records
wave1_orders_dedup = (
    valid_orders_w1
    .dropDuplicates()
)

# Check remaining repeated customer IDs
remaining_customer_duplicates = (
    wave1_customers_dedup
    .groupBy("customer_id")
    .count()
    .filter(F.col("count") > 1)
)

# Check remaining repeated order IDs
remaining_order_duplicates = (
    wave1_orders_dedup
    .groupBy("order_id")
    .count()
    .filter(F.col("count") > 1)
)

print("Customers after exact deduplication:",
      wave1_customers_dedup.count())

print("Orders after exact deduplication:",
      wave1_orders_dedup.count())

print("Remaining repeated customer IDs:",
      remaining_customer_duplicates.count())

print("Remaining repeated order IDs:",
      remaining_order_duplicates.count())

display(remaining_customer_duplicates)
display(remaining_order_duplicates)
```

```python
# Define the unknown member
unknown_customer = spark.createDataFrame(
    [
        (
            -1, "-1",
            "Unknown", "Unknown", "Unknown",
            "Unknown", "Unknown", "Unknown",
            datetime(1900, 1, 1, 0, 0, 0),
            datetime(9999, 12, 31, 0, 0, 0),
            True, False
        )
    ],
    schema=[
        "customer_sk",
        "customer_id",
        "first_name",
        "last_name",
        "email",
        "city",
        "province",
        "loyalty_tier",
        "effective_from",
        "effective_to",
        "is_current",
        "is_inferred"
    ]
)

# Prepare real customer records
real_customers = (
    wave1_customers_dedup
    .withColumn(
        "customer_sk",
        F.row_number().over(
            Window.orderBy("customer_id")
        )
    )
    .withColumn(
        "effective_from",
        F.lit(datetime(1900, 1, 1, 0, 0, 0))
    )
    .withColumn(
        "effective_to",
        F.lit(datetime(9999, 12, 31, 0, 0, 0))
    )
    .withColumn("is_current", F.lit(True))
    .withColumn("is_inferred", F.lit(False))
    .select(
        "customer_sk",
        "customer_id",
        "first_name",
        "last_name",
        "email",
        "city",
        "province",
        "loyalty_tier",
        "effective_from",
        "effective_to",
        "is_current",
        "is_inferred"
    )
)

# Assign positive surrogate keys after the unknown member
real_customers = real_customers.withColumn(
    "customer_sk",
    F.col("customer_sk") + 0
)

# Combine unknown member and real customers
dim_customer = unknown_customer.unionByName(real_customers)

print("Dimension customer rows:", dim_customer.count())
print("Current customer rows:",
      dim_customer.filter(F.col("is_current") == True).count())

display(dim_customer.orderBy("customer_sk").limit(10))
```

```python
# Get the current dimension records, excluding the unknown member
current_customers = dim_customer.filter(
    (F.col("is_current") == True) &
    (F.col("customer_id") != "-1")
)

# Match incoming Wave 2 customers against the current dimension
wave2_customer_status = (
    wave2_customers_dedup.alias("w")
    .join(
        current_customers.select(
            "customer_id",
            "is_inferred"
        ).alias("d"),
        on="customer_id",
        how="left"
    )
)

print(
    "Existing real customers:",
    wave2_customer_status.filter(
        F.col("is_inferred") == False
    ).count()
)

print(
    "Existing inferred customers:",
    wave2_customer_status.filter(
        F.col("is_inferred") == True
    ).count()
)

print(
    "Completely new customers:",
    wave2_customer_status.filter(
        F.col("is_inferred").isNull()
    ).count()
)
```

```python
# Get existing customer records and compare them with Wave 2
existing_customer_comparison = (
    wave2_customers_dedup.alias("w")
    .join(
        current_customers.alias("d"),
        on="customer_id",
        how="inner"
    )
    .select(
        F.col("customer_id"),
        F.col("w.city").alias("new_city"),
        F.col("d.city").alias("old_city"),
        F.col("w.province").alias("new_province"),
        F.col("d.province").alias("old_province"),
        F.col("w.loyalty_tier").alias("new_loyalty_tier"),
        F.col("d.loyalty_tier").alias("old_loyalty_tier"),
        F.col("w.modified_at").alias("incoming_modified_at")
    )
)

# Find customers with at least one Type 2 attribute change
type2_changes = existing_customer_comparison.filter(
    (~F.col("new_city").eqNullSafe(F.col("old_city"))) |
    (~F.col("new_province").eqNullSafe(F.col("old_province"))) |
    (~F.col("new_loyalty_tier").eqNullSafe(F.col("old_loyalty_tier")))
)

print("Existing customers compared:", existing_customer_comparison.count())
print("Customers with Type 2 changes:", type2_changes.count())

display(type2_changes.orderBy("customer_id"))
```

```python
# Get the incoming effective dates for the 20 Type 2 changes
type2_close_rows = (
    wave2_customers_dedup
    .join(
        type2_changes.select("customer_id"),
        on="customer_id",
        how="inner"
    )
    .select("customer_id", "modified_at")
)

# Close only the current dimension version for each affected customer
dim_customer_closed = (
    dim_customer_type1.alias("d")
    .join(
        type2_close_rows.alias("u"),
        on="customer_id",
        how="left"
    )
    .select(
        F.col("d.customer_sk"),
        F.col("d.customer_id"),
        F.col("d.first_name"),
        F.col("d.last_name"),
        F.col("d.email"),
        F.col("d.city"),
        F.col("d.province"),
        F.col("d.loyalty_tier"),
        F.col("d.effective_from"),
        F.when(
            F.col("u.modified_at").isNotNull() & F.col("d.is_current"),
            F.col("u.modified_at")
        ).otherwise(F.col("d.effective_to")).alias("effective_to"),
        F.when(
            F.col("u.modified_at").isNotNull() & F.col("d.is_current"),
            F.lit(False)
        ).otherwise(F.col("d.is_current")).alias("is_current"),
        F.col("d.is_inferred")
    )
)

print(
    "Current records closed:",
    dim_customer_type1.filter(
        F.col("customer_id").isin(
            [r["customer_id"] for r in type2_close_rows.select("customer_id").distinct().collect()]
        ) & (F.col("is_current") == True)
    ).count()
)

print("Dimension rows after closing:", dim_customer_closed.count())
```
